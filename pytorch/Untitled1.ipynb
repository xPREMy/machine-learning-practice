```python
#linear regression using pytorch
import torch
import torch.nn as nn
```

```python
#data declaration
x=torch.tensor([[1],[2],[3],[4],[5]],dtype=torch.float32)
y=torch.tensor([[3],[5],[7.3],[9.5],[11.6]],dtype=torch.float32)
print(x,y)
```

```python
#design an model
class LinearRegression(nn.Module):
  def __init__(self,inputdim,outputdim):
    super(LinearRegression,self).__init__()
    self.lin=nn.Linear(inputdim,outputdim) # does f=w*x+b
  def forward(self,x):
    return self.lin(x)

nfeatures=x.shape[1] # numbers of features
model=LinearRegression(nfeatures,nfeatures)
```

```python
# construction of loss and optimizer
learningrate=0.01
loss=nn.MSELoss()
opt=torch.optim.SGD(model.parameters(),lr=learningrate)
```

```python
#traing loop
n_epochs=1000
for i in range(n_epochs+1):
  ypred=model(x)
  l=loss(y,ypred)
  l.backward()
  opt.step()
  opt.zero_grad()
  w,b=model.parameters()
  print('epochs= ',i,' w= ',w[0][0].item(),' b = ',b.item(),l.item())
```

```python
q=torch.tensor([[1]],dtype=torch.float32)
e=model(q)
print(e)
```
